Desenvolver um programa que leia e apresente quatro notas bimestrais de oito alunos, sem
considerar o cálculo da média.

```python
matriz_notas = []
for aluno in range(8):
    notas_aluno = []
    print(f"\nDigite as notas do Aluno {aluno + 1}:")
    for bimestre in range(4):
        nota = float(input(f"Nota do Bimestre {bimestre + 1}: "))
        notas_aluno.append(nota)
    matriz_notas.append(notas_aluno)
print("\nTabela de Notas:")
print("Aluno      Bimestre 1   Bimestre 2   Bimestre 3   Bimestre 4")
for indice, notas in enumerate(matriz_notas):
    print(f"Aluno {indice + 1}", end="    ")
    for nota in notas:
        print(f"{nota:7.1f}", end="    ")
    print()
```
