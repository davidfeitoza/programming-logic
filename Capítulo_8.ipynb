Desenvolver um programa que leia e apresente quatro notas bimestrais de oito alunos, sem
considerar o cálculo da média.

```python
matriz_notas = []
for aluno in range(8):
    notas_aluno = []
    print(f"\nDigite as notas do Aluno {aluno + 1}:")
    for bimestre in range(4):
        nota = float(input(f"Nota do Bimestre {bimestre + 1}: "))
        notas_aluno.append(nota)
    matriz_notas.append(notas_aluno)
print("\nTabela de Notas:")
print("Aluno      Bimestre 1   Bimestre 2   Bimestre 3   Bimestre 4")
for indice, notas in enumerate(matriz_notas):
    print(f"Aluno {indice + 1}", end="    ")
    for nota in notas:
        print(f"{nota:7.1f}", end="    ")
    print()
```

Desenvolver um programa que simule uma agenda de cadastro pessoal com nome, endereço, código postal, bairro e telefone de dez pessoas. Ao final, o programa deve apresentar seus elementos dispostos em ordem alfabética ascendente a partir do elemento (campo) nome.

Desenvolver um programa que leia os nomes de oito alunos e também suas quatro notas bimestrais. Ao final, deve apresentar o nome de cada aluno classificado em ordem alfabética, bem como suas médias e a média geral dos oito alunos.

```python
alunos = []

for i in range(8):
    nome = input(f"Digite o nome do {i+1}º aluno: ")
    notas = []
    for j in range(4):
        nota = float(input(f"Digite a {j+1}ª nota de {nome}: "))
        notas.append(nota)
    media = sum(notas) / 4 # Calcula a média das 4 notas do aluno
    alunos.append([nome, media]) # Adiciona uma sublista com [nome, média] à lista principal de alunos

# Ordena a lista de alunos em ordem alfabética pelo nome (índice 0 da sublista)
alunos.sort(                      # Chama o método sort() para ordenar a lista 'alunos' "in place" (modifica a lista original)
             key=                 # Define uma função que será usada como critério de ordenação
                 lambda x:        # Define uma função anônima (lambda) que recebe um elemento 'x' da lista
                            x[0]  # A função retorna o primeiro elemento da sublista 'x' (ou seja, o nome do aluno)
          )
# lambda x: x[0] é uma função anônima (sem nome), criada diretamente no lugar onde é usada
# A palavra 'lambda' é usada em Python para criar funções rápidas e simples
# 'x' é o nome dado ao parâmetro da função — neste caso, 'x' representa cada item da lista 'alunos'
# Cada item da lista 'alunos' é uma sublista do tipo: [nome, média] — por exemplo: ["Maria", 7.5]
# O 'x[0]' acessa o primeiro elemento dessa sublista, ou seja, o nome do aluno (posição 0)
# Portanto, essa função serve para dizer ao sort() que a ordenação deve ser feita com base nos nomes dos alunos

# Exibe os alunos com suas médias
soma_medias = 0
print("\nAlunos e suas médias (em ordem alfabética):")
for aluno in alunos:
    print(f"{aluno[0]} - Média: {aluno[1]:.2f}")
    soma_medias += aluno[1] # Adiciona a média do aluno à soma total

# Média geral
media_geral = soma_medias / 8
print(f"\nMédia geral da turma: {media_geral:.2f}")
```

a) Elaborar um programa que leia duas matrizes A e B, cada uma de duas dimensões com cinco linhas e três colunas para valores inteiros. Construir uma matriz C de mesma dimensão, que seja formada pela soma dos elementos da matriz A com os elementos da matriz B. Apresentar os elementos da matriz C.

```python
# Inicializa as matrizes A e B vazias
A = []
B = []

print("Digite os valores da matriz A (5 linhas, 3 colunas):")
for i in range(5):  # Para cada linha
    linha = []
    for j in range(3):  # Para cada coluna
        valor = int(input(f"A[{i}][{j}] = "))
        linha.append(valor)
    A.append(linha)

print("\nDigite os valores da matriz B (5 linhas, 3 colunas):")
for i in range(5):
    linha = []
    for j in range(3):
        valor = int(input(f"B[{i}][{j}] = "))
        linha.append(valor)
    B.append(linha)

# Cria matriz C como soma de A e B
C = []
for i in range(5):
    linha = []
    for j in range(3):
        soma = A[i][j] + B[i][j]
        linha.append(soma)
    C.append(linha)

# Apresenta a matriz C
print("\nMatriz C (soma de A e B):")
for i in range(5):
    for j in range(3):
        print(f"{C[i][j]:4}", end=" ")
    print()  # Pula linha após cada linha da matriz
```

b) Elaborar um programa que leia duas matrizes A e B, cada uma com uma dimensão para sete elementos inteiros. Construir uma matriz C de duas dimensões, cuja primeira coluna deve ser formada pelos elementos da matriz A e a segunda coluna pelos elementos da matriz B. Apresentar a matriz C.

```python
# Ler os 7 elementos da matriz A (lista 1D)
A = []
print("Digite os 7 elementos da matriz A:")
for i in range(7):
    valor = int(input(f"A[{i}] = "))
    A.append(valor)

# Ler os 7 elementos da matriz B (lista 1D)
B = []
print("\nDigite os 7 elementos da matriz B:")
for i in range(7):
    valor = int(input(f"B[{i}] = "))
    B.append(valor)

# Construir a matriz C 2D com 7 linhas e 2 colunas
C = []
for i in range(7):
    linha = [A[i], B[i]]  # primeira coluna A[i], segunda coluna B[i]
    C.append(linha)

# Apresentar a matriz C
print("\nMatriz C:")
for i in range(7):
    print(f"{C[i][0]:4} {C[i][1]:4}")
```

d) Elaborar um programa que leia uma matriz A de uma dimensão com dez elementos intelros. Construir uma matriz C de duas dimensões com três colunas, sendo a primeira coluna da matriz C formada pelos elementos da matriz A somados com 5, a segunda coluna formada pelo valor do cálculo da fatorial de cada elemento correspondente da matriz A e a terceira e última coluna pelos quadrados dos elementos correspondentes da matriz A. Apresentar a matriz C.

```python
# Ler matriz A com 10 elementos inteiros
A = []
print("Digite os 10 elementos da matriz A:")
for i in range(10):
    valor = int(input(f"A[{i}] = "))
    A.append(valor)

# Construir matriz C 10x3
C = []
for i in range(10):
    # calcular fatorial do elemento A[i] sem função
    n = A[i]
    fatorial = 1
    for x in range(2, n+1):
        fatorial *= x

    soma_5 = A[i] + 5
    quad = A[i] ** 2
    C.append([soma_5, fatorial, quad])

# Apresentar matriz C
print("\nMatriz C (colunas: A+5, fatorial(A), quadrado(A)):")
for i in range(10):
    print(f"{C[i][0]:6} {C[i][1]:10} {C[i][2]:8}")
```

Construir uma matriz C de duas dimensões, sendo a primeira coluna da matriz C formada pelos elementos da matriz A multiplicados por 2 e a segunda coluna formada pelos elementos da matriz B subtraidos de 5. f) Elaborar um programa que leia uma matriz A de duas dimensões com cinco linhas e quatro colunas. Construir uma matriz B de mesma dimensão, em que cada elemento seja o fatorial de cada elemento correspondente armazenado na matriz A. Apresentar ao final as matrizes A e B.

```python
# Ler matriz A 5x4
A = []
print("Digite os elementos da matriz A (5 linhas, 4 colunas):")
for i in range(5):
    linha = []
    for j in range(4):
        valor = int(input(f"A[{i}][{j}] = "))
        linha.append(valor)
    A.append(linha)

# Construir matriz B 5x4 com fatoriais
B = []
for i in range(5):
    linha = []
    for j in range(4):
        n = A[i][j]
        fatorial = 1
        # calcular fatorial sem função
        for x in range(2, n+1):
            fatorial *= x
        linha.append(fatorial)
    B.append(linha)

# Apresentar matriz A
print("\nMatriz A:")
for i in range(5):
    for j in range(4):
        print(f"{A[i][j]:4}", end=" ")
    print()

# Apresentar matriz B
print("\nMatriz B (fatorial dos elementos de A):")
for i in range(5):
    for j in range(4):
        print(f"{B[i][j]:8}", end=" ")
    print()
```

g) Elaborar um programa que leia uma matriz A de duas dimensões com quatro linhas e cinco colunas, armazenando nessa matriz os valores das temperaturas em graus Celsius. Construir a matriz B de mesma dimensão, em que cada elemento seja o valor da temperatura em graus Fahrenheit de cada elemento correspondente da matriz A. Apresentar ao final as matrizes A e B.

```python
# Ler matriz A 4x5 com temperaturas em Celsius
A = []
print("Digite as temperaturas em Celsius (matriz 4x5):")
for i in range(4):
    linha = []
    for j in range(5):
        valor = float(input(f"A[{i}][{j}] (°C) = "))
        linha.append(valor)
    A.append(linha)

# Construir matriz B 4x5 com temperaturas em Fahrenheit
B = []
for i in range(4):
    linha = []
    for j in range(5):
        celsius = A[i][j]
        fahrenheit = (celsius * 9/5) + 32
        linha.append(fahrenheit)
    B.append(linha)

# Apresentar matriz A (Celsius)
print("\nMatriz A (Temperaturas em Celsius):")
for i in range(4):
    for j in range(5):
        print(f"{A[i][j]:6.2f}", end=" ")
    print()

# Apresentar matriz B (Fahrenheit)
print("\nMatriz B (Temperaturas em Fahrenheit):")
for i in range(4):
    for j in range(5):
        print(f"{B[i][j]:6.2f}", end=" ")
    print()
```

h) Elaborar um programa que leia uma matriz A do tipo inteira de duas dimensões com cinco linhas e cinco colunas. Construir uma matriz B de mesma dimensão, em que cada elemento seja o dobro de cada elemento correspondente da matriz A, com exceção dos valores situados na diagonal principal (posições B[1, 1] B[2, 2] B[3, 3] B[4, 4] e B[5, 5] ), os quais devem ser o triplo de cada elemento correspondente da matriz A. i) Elaborar um programa que leia uma matriz A do tipo inteira de duas dimensões com sete linhas e sete colunas. Construir a matriz B de mesma dimensão, em que cada elemento seja o somatório de 1 até o valor armazenado na posição da matriz A, com exceção dos valores situados nos indices impares da diagonal principal ( B[1, 1] B[3, 3] B[5, 5] e B[7, 7] ) , os quais devem ser o fatorial de cada elemento correspondente da matriz A. Apresentar ao final a matriz B.

```python
A = []
for i in range(7):
    linha = []
    for j in range(7):
        valor = int(input(f'Digite o valor para A[{i+1},{j+1}]: '))
        linha.append(valor)
    A.append(linha)

B = []
for i in range(7):
    linhaB = []
    for j in range(7):
        if i == j and i % 2 == 0:  # Índices ímpares da diagonal principal (considerando que o problema conta a partir de 1)
            # Cálculo do fatorial
            valor = A[i][j]
            fatorial = 1
            for k in range(1, valor + 1):
                fatorial *= k # fatorial = fatorial * k
            linhaB.append(fatorial)
        else:
            # Cálculo do somatório de 1 até o valor
            valor = A[i][j]
            somatorio = 0
            for k in range(1, valor + 1):
                somatorio += k
            linhaB.append(somatorio)
    B.append(linhaB)

print("\nMatriz B:")
for i in range(7):
    for j in range(7):
        print(f'{B[i][j]:8}', end=' ')
    print()
```

j) Elaborar um programa que leia uma matriz A de duas dimensões com seis linhas e cinco colunas. Construir a matriz B de mesma dimensão, que deve ser formada do seguinte modo: para cada elemento par da matriz A deve ser somado 5 e de cada elemento impar da matriz A deve ser subtraido 4. Apresentar ao final as matrizes A e B.

```python
# Lendo a matriz A de 6 linhas e 5 colunas
A = []
print("Digite os elementos da matriz A (6x5):")
for i in range(6):
    linha = []
    for j in range(5):
        elemento = int(input(f"Elemento A[{i}][{j}]: "))
        linha.append(elemento)
    A.append(linha)

# Construindo a matriz B
B = []
for i in range(6):
    linha_b = []
    for j in range(5):
        elemento_a = A[i][j]
        if elemento_a % 2 == 0:  # Se for par
            elemento_b = elemento_a + 5
        else:  # Se for ímpar
            elemento_b = elemento_a - 4
        linha_b.append(elemento_b)
    B.append(linha_b)

# Apresentando as matrizes A e B
print("\nMatriz A:")
for i in range(6):
    for j in range(5):
        print(f"{A[i][j]:4}", end=" ")
    print()

print("\nMatriz B:")
for i in range(6):
    for j in range(5):
        print(f"{B[i][j]:4}", end=" ")
    print()
```

k) Elaborar um programa que leia uma matriz A do tipo real de duas dimensões com cinco linhas e cinco colunas. Apresentar o somatório dos elementos situados na diagonal principal (posições A[1, 1] A[2, 2] A[3, 3] A[4, 4] e A[5, 5] ) da referida matriz.

```python
# Lendo a matriz A de 5 linhas e 5 colunas
A = []
print("Digite os elementos da matriz A (5x5):")
for i in range(5):
    linha = []
    for j in range(5):
        elemento = float(input(f"Elemento A[{i+1}][{j+1}]: "))
        linha.append(elemento)
    A.append(linha)

# Calculando o somatório da diagonal principal
soma_diagonal = 0
for i in range(5):
    soma_diagonal += A[i][i]  # Elementos onde linha = coluna

# Apresentando a matriz A
print("\nMatriz A:")
for i in range(5):
    for j in range(5):
        print(f"{A[i][j]:8.2f}", end=" ")
    print()

# Apresentando o somatório da diagonal principal
print(f"\nSomatório da diagonal principal: {soma_diagonal:.2f}")
```

l) Elaborar um programa que leia uma matriz A de duas dimensões com 15 linhas e 15 colunas. Apresentar o somatório dos elementos pares situados na diagonal principal da referida matriz.

```python
# Lendo a matriz A de 4 linhas e 4 colunas
A = []
print("Digite os elementos da matriz A (4x4):")
for i in range(4):
    linha = []
    for j in range(4):
        elemento = int(input(f"Elemento A[{i+1}][{j+1}]: "))
        linha.append(elemento)
    A.append(linha)

# Calculando o somatório dos elementos pares na diagonal principal
soma_pares_diagonal = 0
for i in range(4):
    elemento = A[i][i]  # Elemento da diagonal principal
    if elemento % 2 == 0:  # Verifica se é par
        soma_pares_diagonal += elemento

# Apresentando a matriz A
print("\nMatriz A:")
for i in range(4):
    for j in range(4):
        print(f"{A[i][j]:4}", end=" ")
    print()

# Apresentando o somatório dos elementos pares da diagonal principal
print(f"\nSomatório dos elementos pares da diagonal principal: {soma_pares_diagonal}")
```
