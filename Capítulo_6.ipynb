a) Elaborar um programa que efetue a leitura de dez nomes de pessoas em uma matriz A do tipo vetor e apresentá-los em seguida.

```python
nomes = []
for i in range(10):
    nome = input(f"Digite o {i + 1}º nome: ")
    nomes.append(nome)
print("\nNomes digitados:")
for nome in nomes:
  print(nome)
```

b) Elaborar um programa que leia oito elementos inteiros em uma matriz A do tipo vetor. Construir uma matriz B de mesma dimensão com os elementos da matriz A multiplicados por 3. O elemento B[1] deve ser implicado pelo elemento A[1] * 3, o elemento B[2] implicado pelo elemento A[2] * 3, e assim por diante, até 8. Apresentar a matriz B.

```python
A = []
B = []
for i in range(8):
  valor = int(input(f"Digite o valor {i+1} para a matriz A: "))
  A.append(valor)
for valor in A:
  B.append(valor * 3)
print("\nMatriz B (elementos de A multiplicados por 3):")
print(B)
```

c) Escrever um programa que leia duas matrizes (denominadas A e B) do tipo vetor com 5 elementos reais. Construir uma matriz C, sendo cada elemento da matriz C a subtração de um elemento correspondente da matriz A com um elemento correspondente da matriz B, ou seja, a operação de processamento deve estar baseada na operação C[I]=A[I]-B[I]. Ao final, apresentar os elementos da matriz C.

```python
# Lê as matrizes A e B
print("Digite os 5 números da matriz A:")
A = [float(input()) for _ in range(5)]

print("\nDigite os 5 números da matriz B:")
B = [float(input()) for _ in range(5)]

# Calcula a matriz C (A - B)
C = [A[i] - B[i] for i in range(5)]

# Mostra o resultado
print("\nMatriz C (A - B):")
print(C)
```
