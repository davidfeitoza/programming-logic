Construir uma matriz B dos mesmos tipo e dimensão da matriz A, sendo cada elemento da matriz B o somatório de 1 até o valor do elemento correspondente armazenado na matriz A.
```python
# Lê os 5 elementos inteiros
A = [int(input(f"Digite o elemento {i+1}: ")) for i in range(5)]

# Calcula o somatório para cada elemento
B = []
for num in A:
    soma = 0
    if num > 0:
        # Calcula o somatório de 1 até num
        soma = num * (num + 1) // 2
    B.append(soma)

# Mostra os resultados
print("\nMatriz B (somatórios):")
print(B)
```

q) Elaborar um programa que leia 15 elementos reais para uma matriz A de uma dimensão do tipo vetor. Construir uma matriz B de mesmos tipo e dimensão, observando a seguinte lei de formação: "todo elemento da matriz A que possuir índice par deve ter seu elemento dividido por 2; caso contrário, o elemento da matriz A deve ser multiplicado por 1.5". Apresentar os elementos da matriz B.

```python
# Lê os 15 elementos reais
A = [float(input(f"Digite o elemento {i+1}: ")) for i in range(15)]

# Constrói a matriz B conforme a regra de índices
B = []
for i, valor in enumerate(A):
    if i % 2 == 0:  # Índices pares (0, 2, 4...)
        B.append(valor / 2)
    else:            # Índices ímpares (1, 3, 5...)
        B.append(valor * 1.5)

# Mostra os resultados
print("\nMatriz B:")
print(B)
```

s) Elaborar um programa que leia duas matrizes A e B de uma dimensão com seis elementos. A matriz A deve aceitar apenas a entrada de valores pares, enquanto a matriz B deve aceitar apenas a entrada de valores ímpares. A entrada das matrizes deve ser validada pelo programa, e não pelo usuário. Construir uma matriz C que seja o resultado da junção das matrizes A e B, de modo que a matriz C contenha 12 elementos. Apresentar os elementos da matriz C.

```python
# Inicialização das matrizes
A = []
B = []

# Leitura da matriz A (apenas valores pares)
print("Digite 6 números PARES para a matriz A:")
for i in range(6):
    while True:
        try:
            valor = int(input(f"A[{i+1}]: "))
            if valor % 2 == 0:  # Verifica se é par
                A.append(valor)
                break
            else:
                print("Erro: Digite apenas números pares!\n")
        except:
            print("Valor inválido. Digite um número inteiro!\n")

# Leitura da matriz B (apenas valores ímpares)
print("\nDigite 6 números ÍMPARES para a matriz B:")
for i in range(6):
    while True:
        try:
            valor = int(input(f"B[{i+1}]: "))
            if valor % 2 != 0:  # Verifica se é ímpar
                B.append(valor)
                break
            else:
                print("Erro: Digite apenas números ímpares!\n")
        except:
            print("Valor inválido. Digite um número inteiro!\n")

# Construção da matriz C (junção de A e B)
C = A + B

# Apresentação dos resultados
print("\nMatriz A (pares):", A)
print("Matriz B (ímpares):", B)
print("Matriz C (junção):", C)
```

t) Escrever um programa que leia duas matrizes A e B de uma dimensão com dez elementos. A matriz A deve aceitar apenas a entrada de valores divisiveis por 2 e 3, enquanto a matriz B deve aceitar apenas a entrada de valores múltiplos de 5. A entrada das matrizes deve ser validada pelo programa, e não pelo usuário. Construir uma matriz C que seja o resultado da junção das matrizes A e B, de modo que contenha 20 elementos. Apresentar os elementos da matriz C.

```python
# Inicialização das matrizes
A = []
B = []

# Leitura da matriz A (valores divisíveis por 2 e 3)
print("Digite 10 números DIVISÍVEIS por 2 e 3 para a matriz A:")
for i in range(10):
    while True:
        try:
            valor = int(input(f"A[{i+1}]: "))
            if valor % 6 == 0:  # Divisível por 2 e 3 = divisível por 6
                A.append(valor)
                break
            else:
                print("Erro: O valor deve ser divisível por 2 e 3!")
        except:
            print("Valor inválido. Digite um número inteiro!")

# Leitura da matriz B (valores múltiplos de 5)
print("\nDigite 10 números MÚLTIPLOS de 5 para a matriz B:")
for i in range(10):
    while True:
        try:
            valor = int(input(f"B[{i+1}]: "))
            if valor % 5 == 0:  # Verifica se é múltiplo de 5
                B.append(valor)
                break
            else:
                print("Erro: O valor deve ser múltiplo de 5!")
        except:
            print("Valor inválido. Digite um número inteiro!")

# Construção da matriz C (junção de A e B)
C = A + B

# Apresentação dos resultados
print("\nMatriz A (divisíveis por 2 e 3):", A)
print("Matriz B (múltiplos de 5):", B)
print("Matriz C (junção):", C)
```

A matriz A deve aceitar apenas a entrada de valores divisíveis por 2 ou 3, enquanto a matriz B deve aceitar apenas a entrada de valores que não sejam múltiplos de 5. v) Construir um programa que leia uma matriz A de uma dimensão do tipo vetor com 30 elementos do tipo inteiro. Ao final do programa, apresentar a quantidade de valores pares e ímpares existentes na referida matriz.

```python
# Inicializa a matriz A com 5 elementos
A = []
print("Digite 5 números inteiros:")
for i in range(5):
    while True:
        try:
            elemento = int(input(f"Elemento {i + 1}: "))
            A.append(elemento)
            break
        except ValueError:
            print("Por favor, digite um número inteiro válido.")

# Conta os números pares e ímpares
pares = 0
impares = 0

for numero in A:
    if numero % 2 == 0:
        pares += 1
    else:
        impares += 1

# Exibe os resultados
print("\nQuantidade de valores pares:", pares)
print("Quantidade de valores ímpares:", impares)
```

w) Elaborar um programa que leia duas matrizes A e B de uma dimensão do tipo vetor com dez elementos inteiros cada. Construir uma matriz C de mesmos tipo e dimensão que seja formada pelo quadrado da soma dos elementos correspondentes nas matrizes A e B. Apresentar os elementos da matriz C.

```python
# Inicializando as matrizes
A = []
B = []

# Lendo a matriz A
print("Digite os elementos da matriz A (10 números inteiros):")
for i in range(10):
    while True:
        try:
            valor = int(input(f"A[{i + 1}]: "))
            A.append(valor)
            break
        except ValueError:
            print("Por favor, digite um número inteiro válido.")

# Lendo a matriz B
print("\nDigite os elementos da matriz B (10 números inteiros):")
for i in range(10):
    while True:
        try:
            valor = int(input(f"B[{i + 1}]: "))
            B.append(valor)
            break
        except ValueError:
            print("Por favor, digite um número inteiro válido.")

# Construindo a matriz C
C = []
for i in range(10):
    soma = A[i] + B[i]          # Soma dos elementos correspondentes
    quadrado_soma = soma ** 2   # Quadrado da soma
    C.append(quadrado_soma)     # Adicionando à matriz C

# Apresentando os resultados
print("\nElementos da matriz C:")
for i in range(10):
    print(f"C[{i + 1}] = ({A[i]} + {B[i]})² = {C[i]}")
```

x) Elaborar um programa que leia uma matriz A de uma dimensão com seis elementos do tipo real. Construir uma matriz B, em que cada posição de índice ímpar da matriz B deve ser atribuida com um elemento de índice par existente na matriz A e cada posição de índice par da matriz B deve ser atribuida com um elemento de índice ímpar existente na matriz A. Apresentar os elementos das duas matrizes.

```python
# Lê a matriz A com 6 elementos reais
A = []
print("Digite 6 números reais para a matriz A:")
for i in range(6):
    while True:
        try:
            valor = float(input(f"A[{i + 1}]: "))
            A.append(valor)
            break
        except ValueError:
            print("Por favor, digite um número válido.")

# Constrói a matriz B
B = []
for i in range(6):
    if i % 2 == 0:  # Índices pares de B
        # Pega o próximo elemento (índice ímpar de A)
        B.append(A[i + 1])
    else:  # Índices ímpares de B
        # Pega o elemento anterior (índice par de A)
        B.append(A[i - 1])

# Apresenta os resultados
print("\nMatriz A:", A)
print("Matriz B:", B)
```

z) Elaborar um programa que leia uma matriz A de uma dimensão com dez elementos
numéricos inteiros. Apresentar o total de elementos ímpares existentes na matriz e também o percentual do valor total de números ímpares em relação à quantidade total de elementos armazenados na matriz.

```python
# Lê a matriz A com 10 elementos inteiros
A = []
print("Digite 10 números inteiros para a matriz A:")
for i in range(10):
    while True:
        try:
            elemento = int(input(f"Elemento {i + 1}: "))
            A.append(elemento)
            break
        except ValueError:
            print("Por favor, digite um número inteiro válido.")

# Conta os elementos ímpares
contador_impares = 0
for numero in A:
    if numero % 2 != 0:  # Verifica se o número é ímpar
        contador_impares += 1

# Calcula o percentual de ímpares
percentual_impares = (contador_impares / 10) * 100

# Apresenta os resultados
print("\nMatriz A inserida:")
print(A)
print(f"\nTotal de elementos ímpares na matriz: {contador_impares}")
print(f"Percentual de elementos ímpares: {percentual_impares:.2f}%")
```
