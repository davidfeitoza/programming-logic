### **Exemplos**

Elaborar um programa que leia dez nomes e apresente pelo método de pesquisa sequencial os nomes que porventura estejam armazenados na matriz e que coincidam com o nome de entrada de pesquisa. Além de apresentar o nome, o programa deve indicar em que posição da matriz ele está armazenado. Caso o nome pesquisado não seja encontrado, deve informar que o nome pesquisado não foi localizado.

```python
# Fase 1: Leitura dos 10 nomes
nomes = []
print("Digite 10 nomes:")

for i in range(10):
    nome = input(f"Nome {i+1}: ")
    nomes.append(nome)

# Fase 2: Pesquisa sequencial
continuar_pesquisa = True

while continuar_pesquisa:
    print("\n--- Pesquisa Sequencial ---")
    termo = input("Digite o nome a pesquisar (ou 'sair' para encerrar): ")

    if termo.lower() == 'sair':
        continuar_pesquisa = False
        print("Encerrando o programa...")
        break

    encontrado = False

    # Busca sequencial
    for i in range(len(nomes)):
        if nomes[i] == termo:
            print(f"Nome encontrado na posição {i+1}: {nomes[i]}")
            encontrado = True
            break

    if not encontrado:
        print(f"O nome '{termo}' não foi encontrado na lista.")
```

Elaborar um programa que leia dez nomes e apresente por meio do método de pesquisa binária os nomes que porventura estejam armazenados na matriz e que coincidam com o nome de entrada de pesquisa. Além de apresentar o nome, o programa deve indicar em que posição da matriz ele está armazenado. Caso o nome pesquisado não seja encontrado, deve informar que o nome pesquisado não foi localizado.

```python
# Fase 1: Leitura e ordenação dos 10 nomes
nomes = []
print("Digite 10 nomes:")

for i in range(10):
    nome = input(f"Nome {i+1}: ")
    nomes.append(nome)

# Ordena os nomes em ordem alfabética
nomes.sort()

print("\nLista de nomes em ordem alfabética:")
for i, nome in enumerate(nomes):
    print(f"{i+1}. {nome}")

# Fase 2: Pesquisa binária
continuar_pesquisa = True

while continuar_pesquisa:
    print("\n--- Pesquisa Binária ---")
    termo = input("Digite o nome a pesquisar (ou 'sair' para encerrar): ")

    if termo.lower() == 'sair':
        continuar_pesquisa = False
        print("Encerrando o programa...")
        break

    # Inicializa variáveis para busca binária
    esquerda = 0
    direita = len(nomes) - 1
    encontrado = False

    # Executa a busca binária
    while esquerda <= direita and not encontrado:
        meio = (esquerda + direita) // 2

        if nomes[meio] == termo:
            print(f"Nome encontrado na posição {meio+1}: {nomes[meio]}")
            encontrado = True
        elif nomes[meio] < termo:
            esquerda = meio + 1
        else:
            direita = meio - 1

    if not encontrado:
        print(f"O nome '{termo}' não foi encontrado na lista.")
```

b) Elaborar um programa que leia oito elementos numéricos inteiros em uma matriz A de uma dimensão do tipo vetor. Construir uma matriz B de mesmos tipo e dimensão com os elementos da matriz A multiplicados por 5. Montar uma rotina de pesquisa binária para pesquisar os elementos armazenados na matriz B.

```python
# Fase 1: Leitura dos 8 elementos inteiros
A = []
print("Digite 8 números inteiros:")

for i in range(8):
    numero = int(input(f"Elemento {i+1}: "))
    A.append(numero)

# Fase 2: Construção da matriz B (elementos de A multiplicados por 5)
B = [elemento * 5 for elemento in A]

# Ordena a matriz B para permitir pesquisa binária
B.sort()

print("\nMatriz A:", A)
print("Matriz B (A × 5, ordenada):", B)

# Fase 3: Pesquisa binária na matriz B
continuar_pesquisa = True

while continuar_pesquisa:
    print("\n--- Pesquisa Binária na Matriz B ---")
    termo = input("Digite o número a pesquisar (ou 'sair' para encerrar): ")

    if termo.lower() == 'sair':
        continuar_pesquisa = False
        print("Encerrando o programa...")
        break

    try:
        valor = int(termo)
    except ValueError:
        print("Por favor, digite um número inteiro válido.")
        continue

    # Inicializa variáveis para busca binária
    esquerda = 0
    direita = len(B) - 1
    encontrado = False

    # Executa a busca binária
    while esquerda <= direita:
        meio = (esquerda + direita) // 2

        if B[meio] == valor:
            print(f"Valor {valor} encontrado na posição {meio} da matriz B")
            encontrado = True
            break
        elif B[meio] < valor:
            esquerda = meio + 1
        else:
            direita = meio - 1

    if not encontrado:
        print(f"O valor {valor} não foi encontrado na matriz B.")
```

c) Construir um programa que leia 15 elementos numéricos inteiros em uma matriz A de uma dimensão do tipo vetor. Construir uma matriz B de mesmo tipo e dimensão, em que cada elemento seja o fatorial do elemento correspondente armazenado na matriz A. Apresentar os elementos da matriz B ordenados de forma crescente.

```python
# Lê 15 números e armazena na matriz A
A = []
print("Digite 15 números inteiros (de 0 a 20 para evitar fatoriais grandes):")
for i in range(15):
    num = int(input(f"Número {i+1}: "))
    A.append(num)

# Calcula fatorial de cada número e armazena na matriz B
B = []
for num in A:
    fat = 1
    for i in range(1, num + 1):
        fat *= i
    B.append(fat)

# Ordena a matriz B usando o método sort() do Python
B.sort()

# Mostra os resultados
print("\nMatriz B (fatoriais ordenados):")
for num in B:
    print(num)

# Pesquisa binária
while True:
    busca = input("\nDigite um fatorial para buscar (ou 'sair'): ")
    if busca.lower() == 'sair':
        break

    busca = int(busca)
    inicio = 0
    fim = len(B) - 1
    achou = False

    while inicio <= fim:
        meio = (inicio + fim) // 2
        if B[meio] == busca:
            achou = True
            break
        elif B[meio] < busca:
            inicio = meio + 1
        else:
            fim = meio - 1

    if achou:
        print(f"Encontrado na posição {meio}")
    else:
        print("Não encontrado")

print("Programa encerrado.")
```

d) Elaborar um programa que leia uma matriz A com 12 elementos do tipo real. Após a leitura da matriz A, colocar os seus elementos em ordem crescente. Depois, fazer a leitura de uma matriz B também com 12 elementos do tipo real e colocar os elementos em ordem crescente. Construir uma matriz C, em que cada elemento seja a soma do elemento correspondente das matrizes A e B. Colocar em ordem decrescente os elementos da matriz C e apresentar os seus valores.

```python
# Lê a matriz A com 12 elementos reais
print("Digite os 12 elementos da matriz A:")
A = []
for i in range(12):
    elemento = float(input(f"Elemento {i+1}: "))
    A.append(elemento)

# Ordena A em ordem crescente
A.sort()

# Lê a matriz B com 12 elementos reais
print("\nDigite os 12 elementos da matriz B:")
B = []
for i in range(12):
    elemento = float(input(f"Elemento {i+1}: "))
    B.append(elemento)

# Ordena B em ordem crescente
B.sort()

# Cria matriz C com soma dos elementos correspondentes de A e B
C = []
for i in range(12):
    C.append(A[i] + B[i])

# Ordena C em ordem decrescente
C.sort(reverse=True)

# Apresenta os resultados
print("\nMatriz A ordenada crescente:", A)
print("Matriz B ordenada crescente:", B)
print("\nMatriz C (soma A+B) ordenada decrescente:")
for i, valor in enumerate(C):
    print(f"Elemento {i+1}: {valor:.2f}")
```

e) Escrever um programa que leia duas matrizes A e B do tipo vetor com elementos do tipo cadeia, sendo a matriz A com 20 elementos e a matriz B com 30 elementos. Construir uma matriz C, sendo esta a junção das matrizes A e B. Desta forma, a matriz C deve ter a capacidade de armazenar 50 elementos. Apresentar os elementos da matriz C em ordem descendente.

```python
# Lê a matriz A com 20 elementos
print("Digite os 20 elementos da matriz A (texto):")
A = []
for i in range(20):
    elemento = input(f"Elemento A[{i+1}]: ")
    A.append(elemento)

# Lê a matriz B com 30 elementos
print("\nDigite os 30 elementos da matriz B (texto):")
B = []
for i in range(30):
    elemento = input(f"Elemento B[{i+1}]: ")
    B.append(elemento)

# Combina as matrizes A e B na matriz C
C = A + B  # Concatena as duas listas

# Ordena a matriz C em ordem descendente
C.sort(reverse=True)  # Ordem descendente

# Exibe os elementos da matriz C ordenados
print("\nMatriz C (A + B) ordenada em ordem descendente:", C)
```

f) Elaborar um programa que leia 30 elementos numéricos reais em uma matriz A do tipo vetor. Construir uma matriz B de mesmo tipo, observando a seguinte lei de formação: todo elemento da matriz B deve ser o cubo do elemento correspondente da matriz A. Montar o trecho de pesquisa sequencial para pesquisar os elementos armazenados na matriz B.

```python
# Lê 30 números reais
A = [float(input(f"Digite o {i+1}º número: ")) for i in range(30)]

# Calcula o cubo de cada número
B = [num * num * num for num in A]

# Mostra os cubos calculados para referência
print("\nCubos calculados:")
for i, cubo in enumerate(B):
    print(f"{i+1}: {cubo}")

# Pesquisa sequencial
while True:
    busca = input("\nDigite um valor para buscar (ou 'sair'): ")
    if busca.lower() == 'sair':
        break

    try:
        valor = float(busca)
        encontrado = False

        for i in range(30):
            if B[i] == valor:
                print(f"Encontrado na posição {i+1} (número original: {A[i]})")
                encontrado = True

        if not encontrado:
            print(f"Valor {valor} não encontrado!")
            print(f"Exemplo: Se digitou {A[0]} em A, busque {B[0]} em B")

    except:
        print("Digite um número válido!")

print("Programa encerrado.")
```
