Elaborar um programa que leia dez nomes e apresente pelo método de pesquisa sequencial os nomes que porventura estejam armazenados na matriz e que coincidam com o nome de entrada de pesquisa. Além de apresentar o nome, o programa deve indicar em que posição da matriz ele está armazenado. Caso o nome pesquisado não seja encontrado, deve informar que o nome pesquisado não foi localizado.

```python
# Fase 1: Leitura dos 10 nomes
nomes = []
print("Digite 10 nomes:")

for i in range(10):
    nome = input(f"Nome {i+1}: ")
    nomes.append(nome)

# Fase 2: Pesquisa sequencial
continuar_pesquisa = True

while continuar_pesquisa:
    print("\n--- Pesquisa Sequencial ---")
    termo = input("Digite o nome a pesquisar (ou 'sair' para encerrar): ")

    if termo.lower() == 'sair':
        continuar_pesquisa = False
        print("Encerrando o programa...")
        break

    encontrado = False

    # Busca sequencial
    for i in range(len(nomes)):
        if nomes[i] == termo:
            print(f"Nome encontrado na posição {i+1}: {nomes[i]}")
            encontrado = True
            break

    if not encontrado:
        print(f"O nome '{termo}' não foi encontrado na lista.")
```

Elaborar um programa que leia dez nomes e apresente por meio do método de pesquisa binária os nomes que porventura estejam armazenados na matriz e que coincidam com o nome de entrada de pesquisa. Além de apresentar o nome, o programa deve indicar em que posição da matriz ele está armazenado. Caso o nome pesquisado não seja encontrado, deve informar que o nome pesquisado não foi localizado.

```python
# Fase 1: Leitura e ordenação dos 10 nomes
nomes = []
print("Digite 10 nomes:")

for i in range(10):
    nome = input(f"Nome {i+1}: ")
    nomes.append(nome)

# Ordena os nomes em ordem alfabética
nomes.sort()

print("\nLista de nomes em ordem alfabética:")
for i, nome in enumerate(nomes):
    print(f"{i+1}. {nome}")

# Fase 2: Pesquisa binária
continuar_pesquisa = True

while continuar_pesquisa:
    print("\n--- Pesquisa Binária ---")
    termo = input("Digite o nome a pesquisar (ou 'sair' para encerrar): ")

    if termo.lower() == 'sair':
        continuar_pesquisa = False
        print("Encerrando o programa...")
        break

    # Inicializa variáveis para busca binária
    esquerda = 0
    direita = len(nomes) - 1
    encontrado = False

    # Executa a busca binária
    while esquerda <= direita and not encontrado:
        meio = (esquerda + direita) // 2

        if nomes[meio] == termo:
            print(f"Nome encontrado na posição {meio+1}: {nomes[meio]}")
            encontrado = True
        elif nomes[meio] < termo:
            esquerda = meio + 1
        else:
            direita = meio - 1

    if not encontrado:
        print(f"O nome '{termo}' não foi encontrado na lista.")
```

b) Elaborar um programa que leia oito elementos numéricos inteiros em uma matriz A de uma dimensão do tipo vetor. Construir uma matriz B de mesmos tipo e dimensão com os elementos da matriz A multiplicados por 5. Montar uma rotina de pesquisa binária para pesquisar os elementos armazenados na matriz B.

```python
# Fase 1: Leitura dos 8 elementos inteiros
A = []
print("Digite 8 números inteiros:")

for i in range(8):
    numero = int(input(f"Elemento {i+1}: "))
    A.append(numero)

# Fase 2: Construção da matriz B (elementos de A multiplicados por 5)
B = [elemento * 5 for elemento in A]

# Ordena a matriz B para permitir pesquisa binária
B.sort()

print("\nMatriz A:", A)
print("Matriz B (A × 5, ordenada):", B)

# Fase 3: Pesquisa binária na matriz B
continuar_pesquisa = True

while continuar_pesquisa:
    print("\n--- Pesquisa Binária na Matriz B ---")
    termo = input("Digite o número a pesquisar (ou 'sair' para encerrar): ")

    if termo.lower() == 'sair':
        continuar_pesquisa = False
        print("Encerrando o programa...")
        break

    try:
        valor = int(termo)
    except ValueError:
        print("Por favor, digite um número inteiro válido.")
        continue

    # Inicializa variáveis para busca binária
    esquerda = 0
    direita = len(B) - 1
    encontrado = False

    # Executa a busca binária
    while esquerda <= direita:
        meio = (esquerda + direita) // 2

        if B[meio] == valor:
            print(f"Valor {valor} encontrado na posição {meio} da matriz B")
            encontrado = True
            break
        elif B[meio] < valor:
            esquerda = meio + 1
        else:
            direita = meio - 1

    if not encontrado:
        print(f"O valor {valor} não foi encontrado na matriz B.")
```

c) Construir um programa que leia 15 elementos numéricos inteiros em uma matriz A de uma dimensão do tipo vetor. Construir uma matriz B de mesmo tipo e dimensão, em que cada elemento seja o fatorial do elemento correspondente armazenado na matriz A. Apresentar os elementos da matriz B ordenados de forma crescente.

```python
# Lê 15 números e armazena na matriz A
A = []
print("Digite 15 números inteiros (de 0 a 20 para evitar fatoriais grandes):")
for i in range(15):
    num = int(input(f"Número {i+1}: "))
    A.append(num)

# Calcula fatorial de cada número e armazena na matriz B
B = []
for num in A:
    fat = 1
    for i in range(1, num + 1):
        fat *= i
    B.append(fat)

# Ordena a matriz B usando o método sort() do Python
B.sort()

# Mostra os resultados
print("\nMatriz B (fatoriais ordenados):")
for num in B:
    print(num)

# Pesquisa binária
while True:
    busca = input("\nDigite um fatorial para buscar (ou 'sair'): ")
    if busca.lower() == 'sair':
        break

    busca = int(busca)
    inicio = 0
    fim = len(B) - 1
    achou = False

    while inicio <= fim:
        meio = (inicio + fim) // 2
        if B[meio] == busca:
            achou = True
            break
        elif B[meio] < busca:
            inicio = meio + 1
        else:
            fim = meio - 1

    if achou:
        print(f"Encontrado na posição {meio}")
    else:
        print("Não encontrado")

print("Programa encerrado.")
```

g) Elaborar um programa que leia 20 elementos numéricos inteiros em uma matriz A do tipo vetor. Construir uma matriz B de mesma dimensão com os mesmos elementos da matriz A, acrescidos de 2. Colocar os elementos da matriz B em ordem crescente. Montar um trecho de pesquisa binária para pesquisar os elementos armazenados na matriz B.

```python
# Leitura dos 20 elementos inteiros da matriz A
print("Digite 20 números inteiros:")
A = []
for i in range(20):
    while True:
        try:
            num = int(input(f"Elemento {i+1}: "))
            A.append(num)
            break
        except ValueError:
            print("Por favor, digite um número inteiro válido.")

# Construção da matriz B (elementos de A + 2)
B = [num + 2 for num in A]

# Ordenação da matriz B em ordem crescente
B.sort()

print("\nMatriz A (original):", A)
print("Matriz B (A + 2, ordenada):", B)

# Pesquisa binária na matriz B ordenada
while True:
    print("\n--- Pesquisa Binária ---")
    termo = input("Digite o número a pesquisar (ou 'sair' para encerrar): ")

    if termo.lower() == 'sair':
        print("Encerrando o programa...")
        break

    try:
        valor = int(termo)
    except ValueError:
        print("Por favor, digite um número inteiro válido.")
        continue

    # Algoritmo de busca binária
    inicio = 0
    fim = len(B) - 1
    encontrado = False

    while inicio <= fim:
        meio = (inicio + fim) // 2

        if B[meio] == valor:
            encontrado = True
            # Verifica se há ocorrências repetidas
            primeira_pos = meio
            ultima_pos = meio

            # Encontra a primeira ocorrência
            temp = meio - 1
            while temp >= 0 and B[temp] == valor:
                primeira_pos = temp
                temp -= 1

            # Encontra a última ocorrência
            temp = meio + 1
            while temp < len(B) and B[temp] == valor:
                ultima_pos = temp
                temp += 1

            if primeira_pos == ultima_pos:
                print(f"Valor {valor} encontrado na posição {primeira_pos + 1}")
            else:
                print(f"Valor {valor} encontrado nas posições de {primeira_pos + 1} a {ultima_pos + 1}")
            break
        elif B[meio] < valor:
            inicio = meio + 1
        else:
            fim = meio - 1

    if not encontrado:
        print(f"Valor {valor} não encontrado na matriz B.")
```

j) Elaborar um programa que leia duas matrizes A e B do tipo vetor com 15 elementos inteiros cada. Construir duas outras matrizes C e D de mesmo tipo. Cada elemento da matriz C deve ser o somatório do elemento correspondente da matriz A, e cada elemento da matriz D deve ser o fatorial do elemento correspondente da matriz B. Em seguida, construir uma matriz E, que deve conter a diferença dos elementos das matrizes C e D com a soma dos elementos das matrizes A e B. Apresentar os elementos da matriz E em ordem crescente.

```python
# Declaração das matrizes
A = []
B = []
C = []
D = []
E = []

# Leitura dos elementos de A
print("Digite os 15 elementos inteiros da matriz A:")
for i in range(15):
    num = int(input(f"Elemento A[{i+1}]: "))
    A.append(num)

# Leitura dos elementos de B
print("\nDigite os 15 elementos inteiros da matriz B:")
for i in range(15):
    num = int(input(f"Elemento B[{i+1}]: "))
    B.append(num)

# Construção da matriz C (somatório acumulado de A)
soma_acumulada = 0
for num in A:
    soma_acumulada += num
    C.append(soma_acumulada)

# Construção da matriz D (fatorial de cada elemento de B)
for num in B:
    # Cálculo do fatorial (sem função)
    if num < 0:
        fatorial = 0  # Fatorial de negativo não existe, definido como 0
    else:
        fatorial = 1
        for i in range(1, num + 1):
            fatorial *= i
    D.append(fatorial)

# Construção da matriz E: (C[i] - D[i]) + (A[i] + B[i])
for i in range(15):
    elemento_E = (C[i] - D[i]) + (A[i] + B[i])
    E.append(elemento_E)

# Ordenação da matriz E em ordem crescente
E.sort()

# Exibição dos resultados
print("\nMatriz E em ordem crescente:")
for num in E:
    print(num, end=" ")
```

n) Elaborar um programa que leia três matrizes A, B e C de uma dimensão do tipo vetor com 15 elementos inteiros cada. Construir uma matriz D de mesmos tipo e dimensão que seja formada pela soma dos elementos correspondentes às matrizes A, B e C. Montar o trecho de pesquisa binária para pesquisar os elementos existentes na matriz D.

```python
# Declaração das matrizes
A = []
B = []
C = []
D = []

# Leitura dos elementos de A
print("Digite os 5 elementos inteiros da matriz A:")
for i in range(5):
    num = int(input(f"Elemento A[{i+1}]: "))
    A.append(num)

# Leitura dos elementos de B
print("\nDigite os 5 elementos inteiros da matriz B:")
for i in range(5):
    num = int(input(f"Elemento B[{i+1}]: "))
    B.append(num)

# Leitura dos elementos de C
print("\nDigite os 5 elementos inteiros da matriz C:")
for i in range(5):
    num = int(input(f"Elemento C[{i+1}]: "))
    C.append(num)

# Construção da matriz D (soma de A, B e C)
for i in range(5):
    soma = A[i] + B[i] + C[i]
    D.append(soma)

# Ordenação de D para permitir pesquisa binária
D.sort()

# Exibição da matriz D ordenada
print("\nMatriz D ordenada (soma de A, B e C):")
for num in D:
    print(num, end=" ")

# Trecho de pesquisa binária
print("\n\nPesquisa Binária na Matriz D:")
alvo = int(input("Digite o valor inteiro a ser pesquisado em D: "))

# Inicialização dos índices
esquerda = 0
direita = len(D) - 1
encontrado = False

# Algoritmo de pesquisa binária
while esquerda <= direita:
    meio = (esquerda + direita) // 2
    if D[meio] == alvo:
        encontrado = True
        # Verifica se há ocorrências repetidas
        inicio = meio
        fim = meio
        while inicio >= 0 and D[inicio] == alvo:
            inicio -= 1
        while fim < len(D) and D[fim] == alvo:
            fim += 1
        indices = list(range(inicio + 1, fim))
        break
    elif D[meio] < alvo:
        esquerda = meio + 1
    else:
        direita = meio - 1

# Exibição do resultado
if encontrado:
    print(f"Valor {alvo} encontrado nos índices: {indices}")
else:
    print(f"Valor {alvo} não encontrado na matriz D.")
```

```python
# Declaração da matriz A
A = []

# Leitura dos 15 elementos inteiros
print("Digite 15 números inteiros para a matriz A:")
for i in range(15):
    num = int(input(f"Elemento {i+1}: "))
    A.append(num)

# Ordenação da matriz A (necessária para pesquisa binária)
A.sort()

# Exibição da matriz ordenada
print("\nMatriz A ordenada:")
for i, num in enumerate(A):
    print(f"A[{i}] = {num}")

# Loop principal de pesquisa
while True:
    print("\nMenu:")
    print("1. Pesquisar valor na matriz A")
    print("2. Sair")
    opcao = input("Escolha uma opção: ")

    if opcao == '1':
        valor = int(input("Digite o valor a pesquisar: "))

        # Implementação da pesquisa binária diretamente no código
        esquerda = 0
        direita = len(A) - 1
        encontrado = False
        posicao = -1

        while esquerda <= direita and not encontrado:
            meio = (esquerda + direita) // 2
            if A[meio] == valor:
                encontrado = True
                posicao = meio
            elif A[meio] < valor:
                esquerda = meio + 1
            else:
                direita = meio - 1

        if encontrado:
            print(f"Valor {valor} encontrado na posição {posicao}.")
        else:
            print(f"Valor {valor} não encontrado na matriz.")

    elif opcao == '2':
        print("Encerrando o programa...")
        break

    else:
        print("Opção inválida. Tente novamente.")
```

p) Elaborar um programa que leia uma matriz A com dez elementos do tipo cadeia. Construir uma matriz B de mesmos tipo e dimensão que a matriz A. O último elemento da matriz A deve ser o primeiro da matriz B, o penúltimo elemento da matriz A deve ser o segundo da matriz B até que o primeiro elemento da matriz A seja o último da matriz B. Apresentar os elementos da matriz B de forma ordenada ascendente.

```python
# Declaração das matrizes
A = []
B = []

# Leitura dos 10 elementos do tipo cadeia (strings)
print("Digite 10 elementos do tipo cadeia para a matriz A:")
for i in range(10):
    elemento = input(f"Elemento A[{i+1}]: ")
    A.append(elemento)

# Construção da matriz B (invertendo a ordem de A)
for i in range(9, -1, -1):  # Começa do último elemento (índice 9) até o primeiro (índice 0)
    B.append(A[i])

# Ordenação ascendente da matriz B
for i in range(len(B)):
    for j in range(i + 1, len(B)):
        if B[i] > B[j]:
            # Troca os elementos se estiverem fora de ordem
            temp = B[i]
            B[i] = B[j]
            B[j] = temp

# Exibição dos elementos da matriz B ordenados
print("\nMatriz B ordenada de forma ascendente:")
for elemento in B:
    print(elemento)
```

q) Elaborar um programa que leia dez elementos do tipo cadeia em uma matriz A e apresentá-los utilizando pesquisa binária.

```python
# 1. Leitura dos 10 elementos tipo cadeia (strings)
A = []
print("Digite 10 palavras:")
for i in range(10):
    elemento = input(f"{i+1}ª palavra: ")
    A.append(elemento)

# 2. Ordena a lista para aplicar a busca binária
A.sort()
print("\nLista ordenada:", A)

# 3. Leitura do item a ser buscado
chave = input("\nDigite a palavra que deseja buscar: ")

# 4. Implementação da busca binária
inicio = 0
fim = len(A) - 1
encontrado = False

while inicio <= fim:
    meio = (inicio + fim) // 2
    if A[meio] == chave:
        print(f"\nPalavra '{chave}' encontrada na posição {meio} da lista ordenada.")
        encontrado = True
        break
    elif chave < A[meio]:
        fim = meio - 1
    else:
        inicio = meio + 1

if not encontrado:
    print(f"\nPalavra '{chave}' não encontrada na lista.")
```

r) Elaborar um programa que efetue a leitura de dados em duas matrizes (A e B) de uma dimensão do tipo vetor, sendo a matriz A com dez elementos e a matriz B com cinco elementos. Os elementos a serem armazenados nas matrizes devem ser do tipo cadeia. Construir uma matriz C com a capacidade de armazenar um total de 15 elementos e executar a junção das matrizes A e B na matriz C. Apresentar os dados da matriz C em ordem alfabética descendente.

```python
# 1. Leitura da matriz A (10 elementos)
A = []
print("Digite 10 palavras para a matriz A:")
for i in range(10):
    elemento = input(f"A[{i}]: ")
    A.append(elemento)

# 2. Leitura da matriz B (5 elementos)
B = []
print("\nDigite 5 palavras para a matriz B:")
for i in range(5):
    elemento = input(f"B[{i}]: ")
    B.append(elemento)

# 3. Junção das duas listas na matriz C
C = A + B

# 4. Ordenação decrescente (ordem alfabética inversa)
C.sort(reverse=True)

# 5. Exibição da matriz C ordenada
print("\nMatriz C (A + B) em ordem alfabética decrescente:")
for i in range(len(C)):
    print(f"C[{i}] = {C[i]}")
```

s) Elaborar um programa que leia dez elementos numéricos reais em uma matriz A do tipo vetor e apresentar esses elementos por meio de pesquisa sequencial.

```python
# Leitura dos 10 números reais
A = []
print("Digite 10 números reais para a matriz A:")
for i in range(10):
    valor = float(input(f"A[{i}]: "))
    A.append(valor)

# Leitura do valor a ser buscado
busca = float(input("\nDigite o número que deseja buscar na matriz A: "))

# Pesquisa sequencial (linear)
encontrado = False
for i in range(len(A)):
    if A[i] == busca:
        print(f"\nNúmero {busca} encontrado na posição {i} da matriz A.")
        encontrado = True
        break

if not encontrado:
    print(f"\nNúmero {busca} não encontrado na matriz A.")
```
