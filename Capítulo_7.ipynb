### **Pesquisa Sequencial**

1. Iniciar um laço e pedir a leitura de dez nomes.
2. Criar um laço que faça a pesquisa sequencial enquanto o usuário desejar. Na fase de pesquisa deve ser solicitada a informação a ser pesquisada, a qual deve ser comparada com o primeiro elemento. Sendo igual, mostra; caso contrário, avança para o próximo. Se não achar em toda lista o conteúdo pesquisado, informar que não existe o elemento pesquisado; se existir, deve ser mostrado.
3. Encerrar a pesquisa e o programa, quando desejado.

```python
# Lista para armazenar os nomes
nome = [""] * 10  # Equivalente ao conjunto[1..10] de cadeia

# Leitura dos 10 nomes
print("Digite dez nomes:")
for i in range(10):  # i de 0 a 9 (equivalente a 1 a 10 em Portugol)
    nome[i] = input(f"Nome {i+1}: ").strip()

# Loop principal do programa
resp = "sim"
while resp.lower() == "sim":  # .lower() para aceitar "Sim", "SIM", etc
    pesq = input("\nEntre o nome a ser pesquisado: ").strip()

    # Pesquisa sequencial
    i = 0  # Começando do índice 0 (equivalente a 1 em Portugol)
    acha = False

    while i < 10 and not acha:
        if pesq == nome[i]:
            acha = True
        else:
            i += 1

    # Mostrar resultados
    if acha:
        # i+1 para mostrar posição começando em 1 (como em Portugol)
        print(f"{pesq} foi localizado na posição {i+1}")
    else:
        print(f"{pesq} não foi localizado")

    # Perguntar se deseja continuar
    resp = input("\nDeseja continuar? (Sim/Não): ").strip()

print("Programa encerrado.")
```

### **Pesquisa Binária**

1. Iniciar um laço, pedir a leitura de dez nomes e colocá-los em ordem alfabética.
2. Criar um laço que faça a pesquisa enquanto o usuário desejar. Durante a fase de pesquisa deve ser solicitada a informação a ser pesquisada, a qual deve ser comparada pelo método de pesquisa binária. Sendo igual, mostra; caso contrário, avança para o próximo. Se não achar em toda a lista, informar que não existe o elemento pesquisado; se existir, deve mostrá-lo.
3. Encerrar a pesquisa, quando desejado.

```python
# Programa de Pesquisa Bináriaa
# Leitura e ordenação dos nomes
NOME = [""] * 10  # Vetor para 10 nomes

# Leitura dos 10 nomes
print("Digite dez nomes:")
for I in range(10):
    NOME[I] = input(f"Nome {I+1}: ").strip()

# Ordenação usando Bubble Sort
for I in range(9):  # De 0 a 8 (equivalente a 1 a 9)
    for J in range(I+1, 10):  # De I+1 a 9 (equivalente a i+1 a 10)
        if NOME[I] > NOME[J]:
            # Troca os elementos
            X = NOME[I]
            NOME[I] = NOME[J]
            NOME[J] = X

# Pesquisa Binária
RESP = "SIM"
while RESP.upper() == "SIM":
    PESQ = input("\nEntre o nome a ser pesquisado: ").strip()

    # Inicialização da pesquisa
    COMEÇO = 0
    FINAL = 9
    ACHA = False

    # Loop de pesquisa
    while COMEÇO <= FINAL and not ACHA:
        MEIO = (COMEÇO + FINAL) // 2

        if PESQ == NOME[MEIO]:
            ACHA = True
        else:
            if PESQ < NOME[MEIO]:
                FINAL = MEIO - 1
            else:
                COMEÇO = MEIO + 1

    # Resultado da pesquisa
    if ACHA:
        # MEIO+1 para mostrar posição começando em 1
        print(f"{PESQ} foi localizado na posição {MEIO+1}")
    else:
        print(f"{PESQ} não foi localizado")

    # Pergunta se deseja continuar
    RESP = input("\nDeseja continuar? (SIM/NÃO): ").strip()

print("\nPrograma encerrado.")
```

```python
# Leitura dos 10 nomes
nomes = []
print("Digite dez nomes:")
for i in range(10):
    nome = input(f"Nome {i+1}: ").strip()
    nomes.append(nome)
nomes.sort() # Ordenação alfabética (usando sort)

# Pesquisa Binária
continuar = 'sim'
while continuar.lower() == 'sim':
    busca = input("\nEntre o nome a ser pesquisado: ").strip()

    # Inicialização da pesquisa binária
    inicio = 0
    fim = len(nomes) - 1
    encontrado = False
    posicao = -1

    # Algoritmo de pesquisa binária
    while inicio <= fim and not encontrado:
        meio = (inicio + fim) // 2

        if nomes[meio] == busca:
            encontrado = True
            posicao = meio
        elif busca < nomes[meio]:
            fim = meio - 1
        else:
            inicio = meio + 1

    # Exibição do resultado
    if encontrado:
        print(f"'{busca}' foi encontrado na posição {posicao + 1}")
    else:
        print(f"'{busca}' não foi encontrado na lista")

    # Pergunta se deseja continuar
    continuar = input("\nDeseja continuar pesquisando? (sim/não): ").strip()

print("\nPrograma encerrado.")
```

b) Elaborar um programa que leia oito elementos numéricos inteiros em uma matriz A de uma dimensão do tipo vetor. Construir uma matriz B de mesmos tipo e dimensão com os elementos da matriz A multiplicados por 5. Montar uma rotina de pesquisa binária para pesquisar os elementos armazenados na matriz B.

```python
# Leitura dos 8 números inteiros para a matriz A
A = []
print("Digite oito números inteiros:")
for i in range(8):
    while True:
        try:
            num = int(input(f"Elemento {i+1}: "))
            A.append(num)
            break
        except ValueError:
            print("Por favor, digite um número inteiro válido.")

# Construção da matriz B (elementos de A multiplicados por 5)
B = [num * 5 for num in A]

# Ordenação da matriz B para pesquisa binária
for i in range(len(B)-1):
    for j in range(i+1, len(B)):
        if B[i] > B[j]:
            B[i], B[j] = B[j], B[i]  # Troca os elementos

# Exibição das matrizes
print("\nMatriz A (original):", A)
print("Matriz B (A * 5, ordenada):", B)

# Interface de pesquisa
print("\nPesquisa Binária na Matriz B")
while True:
    busca = input("\nDigite um número para buscar na matriz B (ou 'sair' para encerrar): ").strip()

    if busca.lower() == 'sair':
        break

    try:
        valor = int(busca)
        encontrado = False
        inicio = 0
        fim = len(B) - 1
        posicao = -1

        # Algoritmo de pesquisa binária
        while inicio <= fim and not encontrado:
            meio = (inicio + fim) // 2

            if B[meio] == valor:
                encontrado = True
                posicao = meio
            elif valor < B[meio]:
                fim = meio - 1
            else:
                inicio = meio + 1

        # Exibição dos resultados
        if encontrado:
            # Encontra o elemento correspondente em A
            valor_original = valor // 5
            posicao_A = -1
            for i in range(len(A)):
                if A[i] == valor_original:
                    posicao_A = i
                    break

            print(f"Valor {valor} encontrado:")
            print(f" - Posição na matriz B: {posicao + 1}")
            print(f" - Elemento correspondente na matriz A: {A[posicao_A]} (posição {posicao_A + 1})")
        else:
            print(f"Valor {valor} não encontrado na matriz B.")

    except ValueError:
        print("Por favor, digite um número inteiro válido ou 'sair'.")

print("\nPrograma encerrado.")
```
