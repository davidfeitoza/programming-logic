Desenvolver um programa de computador que simule as operações básicas de uma calculadora que opere com a entrada de dois valores do tipo real após a escolha da operação a ser executada. O programa deve apresentar uma lista de opções (menu) com as operações matemáticas de adição, subtração, multiplicação e divisão, além de uma opção de saída do programa. Escolhida a opção desejada, deve ser solicitada a entrada de dois valores numéricos para que seja possível executar o processamento escolhido. Após a execução da operação, o programa deve
apresentar o resultado. Depois da execução de qualquer uma das operações de cálculo, o programa deve voltar para o menu de seleção.

```python
def ROTSOMA():
    print("Rotina de Adição")
    A1 = float(input("Entre o 1o. valor: "))
    B1 = float(input("Entre o 2o. valor: "))
    R1 = A1 + B1
    print("O resultado da operação equivale a:", R1)
    print()

def ROTSUBTRACAO():
    print("Rotina de Subtração")
    A2 = float(input("Entre o 1o. valor: "))
    B2 = float(input("Entre o 2o. valor: "))
    R2 = A2 - B2
    print("O resultado da operação equivale a:", R2)
    print()

def ROTMULTIPLICACAO():
    print("Rotina de Multiplicação")
    A3 = float(input("Entre o 1o. valor: "))
    B3 = float(input("Entre o 2o. valor: "))
    R3 = A3 * B3  # Corrigido: no pseudocódigo estava A3 + B3 (erro)
    print("O resultado da operação equivale a:", R3)
    print()

def ROTDIVISAO():
    print("Rotina de Divisão")
    A4 = float(input("Entre o 1o. valor: "))
    B4 = float(input("Entre o 2o. valor: "))

    if B4 == 0:
        print("O resultado da operação equivale a: ERRO")
    else:
        R4 = A4 / B4
        print("O resultado da operação equivale a:", R4)
    print()

# Programa Principal
OPCAO = 0
while OPCAO != 5:
    print("[1] - Adição")
    print("[2] - Subtração")
    print("[3] - Multiplicação")
    print("[4] - Divisão")
    print("[5] - Fim de Programa")

    try:
        OPCAO = int(input("Escolha uma opção: "))

        if OPCAO != 5:
            if OPCAO == 1:
                ROTSOMA()
            elif OPCAO == 2:
                ROTSUBTRACAO()
            elif OPCAO == 3:
                ROTMULTIPLICACAO()
            elif OPCAO == 4:
                ROTDIVISAO()
            else:
                print("Opção inválida - Tente novamente")
                print()

    except ValueError:
        print("Erro: Digite apenas números!")
        print()
    except Exception as e:
        print(f"Erro inesperado: {e}")
        print()

print("Programa encerrado.")
```

Elaborar um programa de computador que efetue a entrada de dez valores do tipo cadeia (dez nomes) e apresente-os em ordem alfabética ascendente, controlando as etapas de entrada, processamento e saída com uso de sub-rotinas

```python
# Constante
FIM = 10

# Variável global
TABELA = [""] * FIM

def ENTRADA():
    print("Digite", FIM, "palavras:")
    for I in range(FIM):
        TABELA[I] = input(f"Palavra {I+1}: ")

def TROCA(A, B):
    X = A
    A = B
    B = X
    return A, B

def ORDENACAO():
    for I in range(FIM - 1):
        for J in range(I + 1, FIM):
            if TABELA[I] > TABELA[J]:
                # Troca os elementos
                temp = TABELA[I]
                TABELA[I] = TABELA[J]
                TABELA[J] = temp

def PROCESSAMENTO():
    ORDENACAO()

def SAIDA():
    print("\nPalavras ordenadas:")
    for I in range(FIM):
        print(f"{I+1}: {TABELA[I]}")

# Programa principal
if __name__ == "__main__":
    ENTRADA()
    PROCESSAMENTO()
    SAIDA()
```

Desenvolver um programa de computador que armazene valores do
tipo inteiro em uma fila de dados que opere com, no máximo, dez
elementos. O programa deve possuir um menu com as opções
entrar na fila, sair da fila, apresentar primeiro elemento da fila,
apresentar todos os elementos da fila, criar fila e sair.

```python
FIM = 10
MATRIZ = [0] * FIM
INICIO = 0

def VAZIA():
    return INICIO == 0

def CHEIA():
    return INICIO == FIM

def ADICIONAR(ELEMENTO):
    global INICIO, MATRIZ
    if CHEIA():
        return False
    else:
        INICIO += 1
        MATRIZ[INICIO - 1] = ELEMENTO
        return True

def RETIRAR():
    global INICIO, MATRIZ
    if VAZIA():
        return False, 0
    else:
        ELEMENTO = MATRIZ[0]
        for I in range(INICIO - 1):
            MATRIZ[I] = MATRIZ[I + 1]
        INICIO -= 1
        return True, ELEMENTO

def CRIAR():
    global INICIO, MATRIZ
    INICIO = 0
    for I in range(FIM):
        MATRIZ[I] = 0

def ENTRADA():
    global INICIO
    try:
        X = int(input())
        if ADICIONAR(X):
            print(f"Elemento {X} foi inserido na posição {INICIO} da fila.")
        else:
            print(f"Impossível inserir o elemento {X} - fila lotada.")
    except ValueError:
        print("Erro: Digite um número inteiro válido.")

def SAIDA():
    sucesso, X = RETIRAR()
    if sucesso:
        print(f"O elemento {X} foi retirado do início da fila.")
    else:
        print("Impossível retirar elementos - fila vazia.")

def ATUAL():
    if not VAZIA():
        print(f"{MATRIZ[0]} primeiro elemento da fila neste momento.")
    else:
        print("Impossível apresentar - fila vazia.")

def EXIBICAO():
    if not VAZIA():
        for I in range(INICIO):
            print(f"Posição: {I + 1} possui o elemento {MATRIZ[I]}.")
    else:
        print("Impossível apresentar - fila vazia.")

# Programa principal
CRIAR()
OPCAO = 0

while OPCAO != 6:
    print("\n[1] - Entrada")
    print("[2] - Saída")
    print("[3] - Apresenta 1o. da fila")
    print("[4] - Apresentar a fila")
    print("[5] - Nova fila")
    print("[6] - Sair")

    try:
        OPCAO = int(input("Escolha uma opção: "))

        if OPCAO == 1:
            print("Digite o elemento a ser inserido:")
            ENTRADA()
        elif OPCAO == 2:
            SAIDA()
        elif OPCAO == 3:
            ATUAL()
        elif OPCAO == 4:
            EXIBICAO()
        elif OPCAO == 5:
            CRIAR()
            print("Nova fila criada.")
        elif OPCAO == 6:
            print("Programa encerrado.")
        else:
            print("Opção inválida. Tente novamente.")

    except ValueError:
        print("Erro: Digite um número válido.")
```

a) Considerando a necessidade de desenvolver uma agenda que contenha nomes, endereços e telefones de dez pessoas, defina a estrutura de registro apropriada e a codificação de um programa que, com o uso de subprogramas, apresente um menu e suas respectivas rotinas para a execução das seguintes etapas: I)Cadastrar os dez registros. II)Pesquisar os dez registros, um de cada vez, pelo campo nome (usar método sequencial). III)Classificar por ordem alfabética os registros cadastrados. IV)Apresentar todos os registros. V)Sair do programa de cadastro.

```python
# Definição da estrutura de registro
registros = []

def cadastrar_registros():
    """Cadastra os dez registros de pessoas"""
    global registros
    registros = []
    print("\n--- CADASTRO DE 10 PESSOAS ---")
    for i in range(10):
        print(f"\nPessoa {i+1}:")
        nome = input("Nome: ")
        endereco = input("Endereço: ")
        telefone = input("Telefone: ")

        registro = {
            'nome': nome,
            'endereco': endereco,
            'telefone': telefone
        }
        registros.append(registro)
    print("\nCadastro concluído!")

def pesquisar_registro():
    """Pesquisa registros pelo nome usando método sequencial"""
    if not registros:
        print("\nNenhum registro cadastrado. Use a opção 1 primeiro.")
        return

    print("\n--- PESQUISA POR NOME ---")
    nome_pesquisa = input("Digite o nome para pesquisa: ")

    encontrados = []
    for registro in registros:
        if registro['nome'].lower() == nome_pesquisa.lower():
            encontrados.append(registro)

    if encontrados:
        print(f"\nForam encontrados {len(encontrados)} registro(s):")
        for i, registro in enumerate(encontrados, 1):
            print(f"\nRegistro {i}:")
            print(f"Nome: {registro['nome']}")
            print(f"Endereço: {registro['endereco']}")
            print(f"Telefone: {registro['telefone']}")
    else:
        print(f"\nNenhum registro encontrado para o nome '{nome_pesquisa}'.")

def classificar_registros():
    """Classifica os registros por ordem alfabética do nome"""
    if not registros:
        print("\nNenhum registro cadastrado. Use a opção 1 primeiro.")
        return

    registros.sort(key=lambda x: x['nome'].lower())
    print("\nRegistros classificados por ordem alfabética!")

def apresentar_registros():
    """Apresenta todos os registros cadastrados"""
    if not registros:
        print("\nNenhum registro cadastrado. Use a opção 1 primeiro.")
        return

    print("\n--- TODOS OS REGISTROS ---")
    for i, registro in enumerate(registros, 1):
        print(f"\nRegistro {i}:")
        print(f"Nome: {registro['nome']}")
        print(f"Endereço: {registro['endereco']}")
        print(f"Telefone: {registro['telefone']}")
    print(f"\nTotal: {len(registros)} registros")

def menu():
    """Exibe o menu principal"""
    print("\n" + "="*50)
    print("AGENDA PESSOAL")
    print("="*50)
    print("I - Cadastrar os dez registros")
    print("II - Pesquisar registros pelo nome")
    print("III - Classificar registros por ordem alfabética")
    print("IV - Apresentar todos os registros")
    print("V - Sair do programa")
    print("="*50)

# Programa principal
def main():
    while True:
        menu()
        opcao = input("\nDigite a opção desejada (I, II, III, IV ou V): ").upper()

        if opcao == 'I':
            cadastrar_registros()
        elif opcao == 'II':
            pesquisar_registro()
        elif opcao == 'III':
            classificar_registros()
        elif opcao == 'IV':
            apresentar_registros()
        elif opcao == 'V':
            print("\nPrograma encerrado. Até mais!")
            break
        else:
            print("\nOpção inválida! Digite I, II, III, IV ou V.")

# Executar o programa
if __name__ == "__main__":
    main()
```

b) Considerando a necessidade de um programa de computador que armazene o nome e as notas bimestrais de 20 alunos do curso de Técnicas de Programação, defina a
estrutura de registro apropriada e a codificação de um programa que, com o uso de subprogramas, apresente um menu e suas respectivas rotinas para a execução das seguintes etapas: I) Cadastrar os 20 registros (após o cadastro, fazer a
classificação por nome). II) Pesquisar os 20 registros, um de cada vez, pelo campo nome (usar o método binário; nessa pesquisa, o programa deve também apresentar a média do aluno e as mensagens “Aprovado”, caso a média seja maior ou igual a 5, ou “Reprovado” para média abaixo de 5). III) Apresentar todos os registros, médias e a mensagem de aprovação ou reprovação. IV) Apresentar apenas os registros e as médias dos alunos aprovados. V) Apresentar apenas os registros e as médias dos alunos reprovados. VI) Sair do programa de cadastro.

```python
alunos = []

class Aluno:
    def __init__(self, nome, notas):
        self.nome = nome
        self.notas = notas

def cadastrar_alunos():
    global alunos
    alunos.clear()
    print("\n--- Cadastro de Alunos ---")
    for i in range(5):
        print(f"\nAluno {i+1}:")
        nome = input("Nome: ")
        notas = []
        for j in range(4):
            nota = float(input(f"Nota {j+1}: "))
            notas.append(nota)
        alunos.append(Aluno(nome, notas))

    # Ordenação por nome usando Bubble Sort
    n = len(alunos)
    for i in range(n-1):
        for j in range(n-1):
            if alunos[j].nome > alunos[j+1].nome:
                alunos[j], alunos[j+1] = alunos[j+1], alunos[j]
    print("\nCadastro concluído e dados ordenados por nome!")

def calcular_media(notas):
    return sum(notas) / len(notas)

def situacao_aluno(media):
    return "Aprovado" if media >= 5 else "Reprovado"

def pesquisa_binaria():
    print("\n--- Pesquisa Binária ---")
    termo = input("Digite o nome para busca: ")

    inicio = 0
    fim = len(alunos) - 1

    while inicio <= fim:
        meio = (inicio + fim) // 2
        if alunos[meio].nome == termo:
            media = calcular_media(alunos[meio].notas)
            situacao = situacao_aluno(media)
            print(f"\nAluno encontrado:")
            print(f"Nome: {alunos[meio].nome}")
            print(f"Média: {media:.1f} - {situacao}")
            return
        elif alunos[meio].nome < termo:
            inicio = meio + 1
        else:
            fim = meio - 1

    print("\nAluno não encontrado!")

def mostrar_todos():
    print("\n--- Todos os Alunos ---")
    for aluno in alunos:
        media = calcular_media(aluno.notas)
        situacao = situacao_aluno(media)
        print(f"Nome: {aluno.nome} | Média: {media:.1f} | Situação: {situacao}")

def mostrar_aprovados():
    print("\n--- Alunos Aprovados ---")
    for aluno in alunos:
        media = calcular_media(aluno.notas)
        if media >= 5:
            print(f"Nome: {aluno.nome} | Média: {media:.1f}")

def mostrar_reprovados():
    print("\n--- Alunos Reprovados ---")
    for aluno in alunos:
        media = calcular_media(aluno.notas)
        if media < 5:
            print(f"Nome: {aluno.nome} | Média: {media:.1f}")

def main():
    while True:
        print("\n" + "="*40)
        print("Menu Principal")
        print("="*40)
        print("1. Cadastrar Alunos")
        print("2. Pesquisar Aluno")
        print("3. Mostrar Todos")
        print("4. Mostrar Aprovados")
        print("5. Mostrar Reprovados")
        print("6. Sair")

        opcao = input("\nEscolha uma opção: ")

        if opcao == '1':
            cadastrar_alunos()
        elif opcao == '2':
            if len(alunos) == 0:
                print("\nNenhum aluno cadastrado!")
            else:
                pesquisa_binaria()
        elif opcao == '3':
            if len(alunos) == 0:
                print("\nNenhum aluno cadastrado!")
            else:
                mostrar_todos()
        elif opcao == '4':
            if len(alunos) == 0:
                print("\nNenhum aluno cadastrado!")
            else:
                mostrar_aprovados()
        elif opcao == '5':
            if len(alunos) == 0:
                print("\nNenhum aluno cadastrado!")
            else:
                mostrar_reprovados()
        elif opcao == '6':
            print("\nSaindo do programa...")
            break
        else:
            print("\nOpção inválida!")

if __name__ == "__main__":
    main()
```

c) Elaborar um programa que armazene o nome e a altura de 15 pessoas com o uso de registros. O programa deve utilizar subprogramas tanto na apresentação do menu como de suas rotinas para a execução das seguintes etapas: I) Cadastrar os 15 registros. II) Apresentar os registros (nome e altura) das pessoas com
1.5 m ou menos. III) Apresentar os registros (nome e altura) das pessoas com
mais de 1.5 m. IV) Apresentar os registros (nome e altura) das pessoas com
mais de 1.5 m e menos de 2 m. V) Apresentar todos os registros com a média extraída de todas as alturas armazenadas. VI) Sair do programa de cadastro.

```python
pessoas = []

class Pessoa:
    def __init__(self, nome, altura):
        self.nome = nome
        self.altura = altura

def mostrar_menu():
    print("\n" + "="*50)
    print("SISTEMA DE CADASTRO DE PESSOAS")
    print("="*50)
    print("1. Cadastrar 5 pessoas")
    print("2. Mostrar pessoas com 1.5m ou menos")
    print("3. Mostrar pessoas com mais de 1.5m")
    print("4. Mostrar pessoas entre 1.5m e 2.0m")
    print("5. Mostrar todos os registros com média")
    print("6. Sair")
    print("="*50)

def cadastrar_pessoas():
    global pessoas
    pessoas.clear()
    print("\n--- CADASTRO DE PESSOAS ---")
    for i in range(5):
        print(f"\nPessoa {i+1}:")
        nome = input("Nome: ")
        while True:
            try:
                altura = float(input("Altura (em metros): "))
                if altura <= 0:
                    print("A altura deve ser maior que zero!")
                    continue
                break
            except ValueError:
                print("Digite um valor numérico válido!")

        pessoas.append(Pessoa(nome, altura))

    print("\nCadastro concluído com sucesso!")

def mostrar_ate_150():
    print("\n--- PESSOAS COM 1.5m OU MENOS ---")
    encontrou = False
    for pessoa in pessoas:
        if pessoa.altura <= 1.5:
            print(f"Nome: {pessoa.nome} | Altura: {pessoa.altura:.2f}m")
            encontrou = True

    if not encontrou:
        print("Nenhuma pessoa encontrada com 1.5m ou menos.")

def mostrar_acima_150():
    print("\n--- PESSOAS COM MAIS DE 1.5m ---")
    encontrou = False
    for pessoa in pessoas:
        if pessoa.altura > 1.5:
            print(f"Nome: {pessoa.nome} | Altura: {pessoa.altura:.2f}m")
            encontrou = True

    if not encontrou:
        print("Nenhuma pessoa encontrada com mais de 1.5m.")

def mostrar_entre_150_e_200():
    print("\n--- PESSOAS ENTRE 1.5m E 2.0m ---")
    encontrou = False
    for pessoa in pessoas:
        if 1.5 < pessoa.altura < 2.0:
            print(f"Nome: {pessoa.nome} | Altura: {pessoa.altura:.2f}m")
            encontrou = True

    if not encontrou:
        print("Nenhuma pessoa encontrada entre 1.5m e 2.0m.")

def calcular_media_alturas():
    if not pessoas:
        return 0
    total = sum(pessoa.altura for pessoa in pessoas)
    return total / len(pessoas)

def mostrar_todos_com_media():
    print("\n--- TODOS OS REGISTROS ---")
    if not pessoas:
        print("Nenhuma pessoa cadastrada.")
        return

    for i, pessoa in enumerate(pessoas, 1):
        print(f"{i}. Nome: {pessoa.nome} | Altura: {pessoa.altura:.2f}m")

    media = calcular_media_alturas()
    print(f"\nMédia das alturas: {media:.2f}m")

def main():
    while True:
        mostrar_menu()
        opcao = input("\nEscolha uma opção (1-6): ")

        if opcao == '1':
            cadastrar_pessoas()
        elif opcao == '2':
            if not pessoas:
                print("\nErro: Cadastre as pessoas primeiro!")
            else:
                mostrar_ate_150()
        elif opcao == '3':
            if not pessoas:
                print("\nErro: Cadastre as pessoas primeiro!")
            else:
                mostrar_acima_150()
        elif opcao == '4':
            if not pessoas:
                print("\nErro: Cadastre as pessoas primeiro!")
            else:
                mostrar_entre_150_e_200()
        elif opcao == '5':
            mostrar_todos_com_media()
        elif opcao == '6':
            print("\nSaindo do programa...")
            print("Obrigado por usar nosso sistema!")
            break
        else:
            print("\nOpção inválida! Digite um número de 1 a 6.")

        input("\nPressione Enter para continuar...")

if __name__ == "__main__":
    main()
```
