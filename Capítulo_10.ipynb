Desenvolver um programa de computador que simule as operações básicas de uma calculadora que opere com a entrada de dois valores do tipo real após a escolha da operação a ser executada. O programa deve apresentar uma lista de opções (menu) com as operações matemáticas de adição, subtração, multiplicação e divisão, além de uma opção de saída do programa. Escolhida a opção desejada, deve ser solicitada a entrada de dois valores numéricos para que seja possível executar o processamento escolhido. Após a execução da operação, o programa deve
apresentar o resultado. Depois da execução de qualquer uma das operações de cálculo, o programa deve voltar para o menu de seleção.

```python
def ROTSOMA():
    print("Rotina de Adição")
    A1 = float(input("Entre o 1o. valor: "))
    B1 = float(input("Entre o 2o. valor: "))
    R1 = A1 + B1
    print("O resultado da operação equivale a:", R1)
    print()

def ROTSUBTRACAO():
    print("Rotina de Subtração")
    A2 = float(input("Entre o 1o. valor: "))
    B2 = float(input("Entre o 2o. valor: "))
    R2 = A2 - B2
    print("O resultado da operação equivale a:", R2)
    print()

def ROTMULTIPLICACAO():
    print("Rotina de Multiplicação")
    A3 = float(input("Entre o 1o. valor: "))
    B3 = float(input("Entre o 2o. valor: "))
    R3 = A3 * B3  # Corrigido: no pseudocódigo estava A3 + B3 (erro)
    print("O resultado da operação equivale a:", R3)
    print()

def ROTDIVISAO():
    print("Rotina de Divisão")
    A4 = float(input("Entre o 1o. valor: "))
    B4 = float(input("Entre o 2o. valor: "))

    if B4 == 0:
        print("O resultado da operação equivale a: ERRO")
    else:
        R4 = A4 / B4
        print("O resultado da operação equivale a:", R4)
    print()

# Programa Principal
OPCAO = 0
while OPCAO != 5:
    print("[1] - Adição")
    print("[2] - Subtração")
    print("[3] - Multiplicação")
    print("[4] - Divisão")
    print("[5] - Fim de Programa")

    try:
        OPCAO = int(input("Escolha uma opção: "))

        if OPCAO != 5:
            if OPCAO == 1:
                ROTSOMA()
            elif OPCAO == 2:
                ROTSUBTRACAO()
            elif OPCAO == 3:
                ROTMULTIPLICACAO()
            elif OPCAO == 4:
                ROTDIVISAO()
            else:
                print("Opção inválida - Tente novamente")
                print()

    except ValueError:
        print("Erro: Digite apenas números!")
        print()
    except Exception as e:
        print(f"Erro inesperado: {e}")
        print()

print("Programa encerrado.")
```

Elaborar um programa de computador que efetue a entrada de dez valores do tipo cadeia (dez nomes) e apresente-os em ordem alfabética ascendente, controlando as etapas de entrada, processamento e saída com uso de sub-rotinas

```python
# Constante
FIM = 10

# Variável global
TABELA = [""] * FIM

def ENTRADA():
    print("Digite", FIM, "palavras:")
    for I in range(FIM):
        TABELA[I] = input(f"Palavra {I+1}: ")

def TROCA(A, B):
    X = A
    A = B
    B = X
    return A, B

def ORDENACAO():
    for I in range(FIM - 1):
        for J in range(I + 1, FIM):
            if TABELA[I] > TABELA[J]:
                # Troca os elementos
                temp = TABELA[I]
                TABELA[I] = TABELA[J]
                TABELA[J] = temp

def PROCESSAMENTO():
    ORDENACAO()

def SAIDA():
    print("\nPalavras ordenadas:")
    for I in range(FIM):
        print(f"{I+1}: {TABELA[I]}")

# Programa principal
if __name__ == "__main__":
    ENTRADA()
    PROCESSAMENTO()
    SAIDA()
```

Desenvolver um programa de computador que armazene valores do
tipo inteiro em uma fila de dados que opere com, no máximo, dez
elementos. O programa deve possuir um menu com as opções
entrar na fila, sair da fila, apresentar primeiro elemento da fila,
apresentar todos os elementos da fila, criar fila e sair.

```python
FIM = 10
MATRIZ = [0] * FIM
INICIO = 0

def VAZIA():
    return INICIO == 0

def CHEIA():
    return INICIO == FIM

def ADICIONAR(ELEMENTO):
    global INICIO, MATRIZ
    if CHEIA():
        return False
    else:
        INICIO += 1
        MATRIZ[INICIO - 1] = ELEMENTO
        return True

def RETIRAR():
    global INICIO, MATRIZ
    if VAZIA():
        return False, 0
    else:
        ELEMENTO = MATRIZ[0]
        for I in range(INICIO - 1):
            MATRIZ[I] = MATRIZ[I + 1]
        INICIO -= 1
        return True, ELEMENTO

def CRIAR():
    global INICIO, MATRIZ
    INICIO = 0
    for I in range(FIM):
        MATRIZ[I] = 0

def ENTRADA():
    global INICIO
    try:
        X = int(input())
        if ADICIONAR(X):
            print(f"Elemento {X} foi inserido na posição {INICIO} da fila.")
        else:
            print(f"Impossível inserir o elemento {X} - fila lotada.")
    except ValueError:
        print("Erro: Digite um número inteiro válido.")

def SAIDA():
    sucesso, X = RETIRAR()
    if sucesso:
        print(f"O elemento {X} foi retirado do início da fila.")
    else:
        print("Impossível retirar elementos - fila vazia.")

def ATUAL():
    if not VAZIA():
        print(f"{MATRIZ[0]} primeiro elemento da fila neste momento.")
    else:
        print("Impossível apresentar - fila vazia.")

def EXIBICAO():
    if not VAZIA():
        for I in range(INICIO):
            print(f"Posição: {I + 1} possui o elemento {MATRIZ[I]}.")
    else:
        print("Impossível apresentar - fila vazia.")

# Programa principal
CRIAR()
OPCAO = 0

while OPCAO != 6:
    print("\n[1] - Entrada")
    print("[2] - Saída")
    print("[3] - Apresenta 1o. da fila")
    print("[4] - Apresentar a fila")
    print("[5] - Nova fila")
    print("[6] - Sair")

    try:
        OPCAO = int(input("Escolha uma opção: "))

        if OPCAO == 1:
            print("Digite o elemento a ser inserido:")
            ENTRADA()
        elif OPCAO == 2:
            SAIDA()
        elif OPCAO == 3:
            ATUAL()
        elif OPCAO == 4:
            EXIBICAO()
        elif OPCAO == 5:
            CRIAR()
            print("Nova fila criada.")
        elif OPCAO == 6:
            print("Programa encerrado.")
        else:
            print("Opção inválida. Tente novamente.")

    except ValueError:
        print("Erro: Digite um número válido.")
```

a) Considerando a necessidade de desenvolver uma agenda que contenha nomes, endereços e telefones de dez pessoas, defina a estrutura de registro apropriada e a codificação de um programa que, com o uso de subprogramas, apresente um menu e suas respectivas rotinas para a execução das seguintes etapas: I)Cadastrar os dez registros. II)Pesquisar os dez registros, um de cada vez, pelo campo nome (usar método sequencial). III)Classificar por ordem alfabética os registros cadastrados. IV)Apresentar todos os registros. V)Sair do programa de cadastro.

```python
# Definição da estrutura de registro
registros = []

def cadastrar_registros():
    """Cadastra os dez registros de pessoas"""
    global registros
    registros = []
    print("\n--- CADASTRO DE 10 PESSOAS ---")
    for i in range(10):
        print(f"\nPessoa {i+1}:")
        nome = input("Nome: ")
        endereco = input("Endereço: ")
        telefone = input("Telefone: ")

        registro = {
            'nome': nome,
            'endereco': endereco,
            'telefone': telefone
        }
        registros.append(registro)
    print("\nCadastro concluído!")

def pesquisar_registro():
    """Pesquisa registros pelo nome usando método sequencial"""
    if not registros:
        print("\nNenhum registro cadastrado. Use a opção 1 primeiro.")
        return

    print("\n--- PESQUISA POR NOME ---")
    nome_pesquisa = input("Digite o nome para pesquisa: ")

    encontrados = []
    for registro in registros:
        if registro['nome'].lower() == nome_pesquisa.lower():
            encontrados.append(registro)

    if encontrados:
        print(f"\nForam encontrados {len(encontrados)} registro(s):")
        for i, registro in enumerate(encontrados, 1):
            print(f"\nRegistro {i}:")
            print(f"Nome: {registro['nome']}")
            print(f"Endereço: {registro['endereco']}")
            print(f"Telefone: {registro['telefone']}")
    else:
        print(f"\nNenhum registro encontrado para o nome '{nome_pesquisa}'.")

def classificar_registros():
    """Classifica os registros por ordem alfabética do nome"""
    if not registros:
        print("\nNenhum registro cadastrado. Use a opção 1 primeiro.")
        return

    registros.sort(key=lambda x: x['nome'].lower())
    print("\nRegistros classificados por ordem alfabética!")

def apresentar_registros():
    """Apresenta todos os registros cadastrados"""
    if not registros:
        print("\nNenhum registro cadastrado. Use a opção 1 primeiro.")
        return

    print("\n--- TODOS OS REGISTROS ---")
    for i, registro in enumerate(registros, 1):
        print(f"\nRegistro {i}:")
        print(f"Nome: {registro['nome']}")
        print(f"Endereço: {registro['endereco']}")
        print(f"Telefone: {registro['telefone']}")
    print(f"\nTotal: {len(registros)} registros")

def menu():
    """Exibe o menu principal"""
    print("\n" + "="*50)
    print("AGENDA PESSOAL")
    print("="*50)
    print("I - Cadastrar os dez registros")
    print("II - Pesquisar registros pelo nome")
    print("III - Classificar registros por ordem alfabética")
    print("IV - Apresentar todos os registros")
    print("V - Sair do programa")
    print("="*50)

# Programa principal
def main():
    while True:
        menu()
        opcao = input("\nDigite a opção desejada (I, II, III, IV ou V): ").upper()

        if opcao == 'I':
            cadastrar_registros()
        elif opcao == 'II':
            pesquisar_registro()
        elif opcao == 'III':
            classificar_registros()
        elif opcao == 'IV':
            apresentar_registros()
        elif opcao == 'V':
            print("\nPrograma encerrado. Até mais!")
            break
        else:
            print("\nOpção inválida! Digite I, II, III, IV ou V.")

# Executar o programa
if __name__ == "__main__":
    main()
```
