### **Desvio Condicional Simples**

1. Definir a entrada de dois valores incógnitos (variáveis A e B). 2. Efetuar a adição dos valores A e B e atribuir o resultado da adição à variável X. 3. Apresentar o resultado da soma armazenada na variável X, caso a variável X tenha seu valor maior que 10.

```python
a = float(input("Digite o valor de a: "))
b = float(input("Digite o valor de b: "))
x = a + b
if x > 10:
  print(f"O resultado da soma é {x}, que é maior que 10.")
```

### **Desvio Condicional Composto**

1. Definir a entrada de dois valores incógnitos (variáveis A e B). 2. Efetuar a adição dos valores A e B e atribuir o resultado da adição à variável X. 3. Verificar se o valor da variável X é maior ou igual a 10; caso seja maior ou igual a 10, proceder ao cálculo de X + 5, atribuindo seu resultado à variável R. Se o valor da variável X não for maior ou igual a 10, proceder ao cálculo de X - 7, atribuindo seu resultado à variável R. 4. Apresentar o resultado da variável R.

```python
A = float(input("Digite o valor de A: "))
B = float(input("Digite o valor de B: "))
X = A + B
if X >= 10:
  R = X + 5
else:
  R = X - 7
print(f"O resultado é {R}.")
```

### **Tomada de Decisão Sequencial**

1. Definir a entrada de um valor numérico inteiro (variável N). 2. Verificar se N = 1 e, se for, apresentar a mensagem "você entrou o valor 1". 3. Verificar se N = 2 e, se for, apresentar a mensagem "você entrou o valor 2". 4. Verificar se N < 1 e, se for, apresentar a mensagem "você entrou valor muito baixo". 5. Verificar se N > 2 e, se for, apresentar a mensagem "você entrou valor muito alto".


```python
N = int(input("Digite um valor inteiro para N: "))
if N == 1:
  print("Você entrou o valor 1")
elif N == 2:
  print("Você entrou o valor 2")
elif N < 1:
  print("Você entrou valor muito baixo")
elif N > 2:
  print("Você entrou valor muito alto")
```

### **Tomada de Decisão Encadeada**

1. Ler o valor de salário atual (variável SA). 2. Verificar se o valor da variável SA é menor que 500. Se sim, reajustar o valor com mais 15%, atribuindo o novo valor à variável NS. Se não, verificar a próxima condição. Note que essa condição estabelece o reajuste de 15% aos salários entre os valores de 0 até 499. O que estiver acima dessa faixa é verificado posteriormente. 3. Verificar se o valor da variável SA é menor ou igual a 1000. Se sim, reajustar o valor com mais 10%, atribuindo o novo valor à variável NS. Essa condição estabelece o reajuste de 10% aos salários entre 500 (após a condição falsa do passo 3) até 1000. O que estiver fora dessa faixa é automaticamente reajustado com mais 5% atribuído à variável NS, pois se trata de valores acima de 1000. 4. Apresentar o valor do novo salário, implicado na variável NS.


```python
SA = float(input("Digite o salário atual: "))
if SA < 500:
  NS = SA * 1.15 # Reajuste de 15%
else:
  if SA <= 1000:
    NS = SA * 1.10 # Reajuste de 10%
  else:
    NS = SA * 1.05 # Reajuste de 5%
print(f"Novo salário: R$ {NS:.2f}")
```

### **Operador Lógico de Conjunção**

1. Efetuar a leitura de um valor numérico inteiro (variável NÚMERO). 2. Verificar se o valor fornecido é maior ou igual a 20 e se o mesmo valor é menor ou igual a 90. Se esta condição for verdadeira, apresentar a mensagem "O número está na faixa de 20 a 90"; caso contrário, apresentar a mensagem "O número está fora da faixa de 20 a 90".


```python
NUMERO = int(input("Digite um número inteiro: "))
if NUMERO >= 20 and NUMERO <= 90:
  print("O número está na faixa de 20 a 90")
else:
  print("O número está fora da faixa de 20 a 90")
```

### **Operador Lógico de Disjunção Inclusiva**

1. Efetuar a entrada do sexo de uma pessoa (variável SEXO). 2. Verificar se o valor fornecido para a variável SEXO é válido, ou seja, se o valor da variável SEXO é igual a F ou M. Se o valor for válido, apresentar a mensagem "Sexo válido"; caso contrário, apresentar a mensagem "Sexo inválido".

```python
SEXO = input("Digite o sexo (F ou M): ").strip().upper() # Remove espaços extras e converte a entrada para maiúsculas
if SEXO == "F" or SEXO == "M":
  print("Sexo válido")
else:
  print("Sexo inválido")
```

### **Operador Lógico de Negação**

1. Efetuar a entrada, respectivamente, dos valores das variáveis A, B e X. 2. Verificar se o valor fornecido para a variável X realmente não é maior que 5. Sendo esta condição verdadeira, processar a operação C← A + B; caso contrário, deve ser realizada a operação C-A-B. 3. Apresentar o resultado obtido na variável C.


```python
A = float(input("Digite o valor de A: "))
B = float(input("Digite o valor de B: "))
X = float(input("Digite o valor de X: "))
if not X > 5: # isto equivale a X <= 5
  C = A + B
else:
  C = A - B
print(f"O resultado obtido na variável C é: {C}")
```

### **Operador Lógico de Disjunção Exclusiva**

1. Efetuar a entrada do nome (variável N1) e do sexo (variável S1) da primeira pessoa. 2. Efetuar a entrada do nome (variável N2) e do sexo (variável S2) da segunda pessoa. 3. Verificar se o sexo da primeira pessoa é exclusivamente igual a um determinado sexo e se o sexo da segunda pessoa é exclusivamente igual ao sexo informado exclusivamente para a primeira pessoa. Se um dos sexos for exclusivamente igual a um dos sexos informados e o outro for diferente, o programa apresenta uma mensagem informando que podem ser formados pares. Caso contrário, o programa apresenta uma mensagem informando que o par não pode ser formado.


```python
N1 = input("Digite o nome da primeira pessoa: ")
S1 = input("Digite o sexo da primeira pessoa (M/F): ").upper()
N2 = input("Digite o nome da segunda pessoa: ")
S2 = input("Digite o sexo da segunda pessoa (M/F): ").upper()
if (S1 == "M") ^ (S2 == "M"): # XOR retorna True se apenas um for "M"
  print(f"{N1} e {N2} podem formar um par.")
else:
  print(f"{N1} e {N2} não podem formar um par.")
```

### **Resto da Divisão**

1. Ler um valor numérico inteiro qualquer (variável N). 2. Calcular o resto da divisão de N por 4, usar a variável R4. 3. Calcular o resto da divisão de N por 5, usar a variável R5. 4. Verificar se as variáveis R4 e R5 possuem o valor de resto igual a zero; se sim, apresentar o valor da variável N; se não, apresentar a mensagem "Valor não é divisivel por 4 e 5".

```python
N = int(input("Digite um número inteiro: "))
R4 = N % 4
R5 = N % 5
if R4 == 0 and R5 == 0:
  print(f"O valor {N} é divisível por 4 e 5.")
else:
  print("Valor não é divisível por 4 e 5.")
```
